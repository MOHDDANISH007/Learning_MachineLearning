```python
import numpy as np
from sklearn.preprocessing import StandardScaler
from scipy.stats import norm
import seaborn as sns

# Step 1: Age data
ages = np.array([52, 55, 56, 58, 59,
                 60, 60, 61, 62, 63,
                 64, 65, 66, 67, 68,
                 69, 70, 71, 72, 74]).reshape(-1, 1)



# occurrence of each age
unique, counts = np.unique(ages, return_counts=True)
print(
    "Age Occurrence:\n",
    dict(zip(unique, counts))
)






# Step 2: Standardize ages
scaler = StandardScaler()
ages_std = scaler.fit_transform(ages)



# Step 3: Standardize the value 72 (find Z-score)
z_72 = scaler.transform([[72]])
z_72 = z_72[0][0]  # convert array → float



# Step 4: Probability age > 72
probability = 1 - norm.cdf(z_72)
print("Probability that age > 72:", round(probability, 4)*100)


plt.figure(figsize=(10, 6))
sns.kdeplot(ages_std, shade=True)
plt.show()
```
