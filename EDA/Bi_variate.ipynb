```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

flight_Data = pd.read_csv('./flights.csv')
resturant_Data = pd.read_csv('./Restaurant_Reviews.csv')
iris_Data = pd.read_csv('./Iris.csv')
titanic_Data = pd.read_csv('./Titanic-Dataset.csv')

titanic_Data.head(20)

# titanic_Data.info()
```

```python
# If we have Numerical Data then how do we visualize it by using EDA or Bi_variate analysis
# here we are using scatter plot to visualize the data...


# 1. Numerical to Numerical Data
# And data is numerical to numerical relationship (bivariate analysis means when we want to analyze the relationship between two variables and the variables are numerical to numerical then we usually use scatter plot)


resturant_Data.head()

# sns.scatterplot(x='bill_amount', y='tip', data=resturant_Data)
sns.scatterplot(
    x='bill_amount',
    y='tip',
    data=resturant_Data,
    hue='smoker',
    style='gender',
    size='gender'
)
```

```python
# 2. Numerical to Categorical Data
# And data is numerical to categorical relationship (bivariate analysis means when we want to analyze the relationship between two variables and the variables are numerical to categorical then we usually use box plot)


# sns.barplot(x=titanic_Data['Pclass'], y=titanic_Data['Age'], hue=titanic_Data['Sex'])
sns.barplot(x=titanic_Data['Pclass'], y=titanic_Data['Fare'], hue=titanic_Data['Sex'])
```

```python
# 2. Numerical to Categorical Data
# And data is numerical to categorical relationship (bivariate analysis means when we want to analyze the relationship between two variables and the variables are numerical to categorical then we usually use box plot)

# box plot is used to detect outliers
sns.boxplot(x=titanic_Data['Pclass'], y=titanic_Data['Age'], hue=titanic_Data['Survived'])
```

```python
# cross tab
pd.crosstab(titanic_Data['Age'], titanic_Data['Sex'])


# heatmap show the relationship between two variables

sns.heatmap(pd.crosstab(titanic_Data['Pclass'], titanic_Data['Survived']))

# Group by Pclass, then take the 'Survived' column, then find the mean
titanic_Data.groupby('Pclass')['Survived'].mean()*100

titanic_Data[titanic_Data['Survived'] == 0].groupby('Pclass')


# cross tab
pd.crosstab(titanic_Data['SibSp'], titanic_Data['Survived'])
```
