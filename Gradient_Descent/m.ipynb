```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# --- Step 1: Data ---
X = np.array([[1, 1],
              [2, 1],
              [1, 2]])
y = np.array([5, 8, 9])

# --- Step 2: Initialize parameters ---
m = np.zeros(X.shape[1])  # slopes [m1, m2]
b = 0                      # intercept
lr = 0.01                  # learning rate
iterations = 1000

# --- Step 3: Predictions BEFORE gradient descent ---
y_pred_initial = X @ m + b

# --- Step 4: Gradient Descent Loop ---
for i in range(iterations):
    y_pred = X @ m + b  
    error = y - y_pred
    # print(f"error: {error.shape}")
    dm = -2 * X.T @ error
    # print(f'dm: {dm}')
    db = -2 * np.sum(error)
    m = m - lr * dm
    b = b - lr * db

# --- Step 5: Predictions AFTER gradient descent ---
y_pred_final = X @ m + b
print(f"Predictions after GD: {y_pred_final}")

# --- Step 6: Plotting 3D graph ---
fig = plt.figure(figsize=(12,6))
ax = fig.add_subplot(121, projection='3d')
ax.scatter(X[:,0], X[:,1], y, color='blue', label='Actual')

# Plane before gradient descent
x1_range = np.linspace(0, 3, 10)
x2_range = np.linspace(0, 3, 10)
x1_grid, x2_grid = np.meshgrid(x1_range, x2_range)
y_plane_initial = m[0]*0 + m[1]*0 + b  # initial m=0, b=0 plane is just z=0
y_plane_initial = np.zeros_like(x1_grid)  # plane at z=0
ax.plot_surface(x1_grid, x2_grid, y_plane_initial, alpha=0.5, color='red')
ax.set_title('Before Gradient Descent')
ax.set_xlabel('x1 (size)')
ax.set_ylabel('x2 (bedrooms)')
ax.set_zlabel('Price')

# --- Step 7: Plot after gradient descent ---
ax2 = fig.add_subplot(122, projection='3d')
ax2.scatter(X[:,0], X[:,1], y, color='blue', label='Actual')

# Plane after gradient descent
y_plane_final = m[0]*x1_grid + m[1]*x2_grid + b
ax2.plot_surface(x1_grid, x2_grid, y_plane_final, alpha=0.5, color='green')
ax2.set_title('After Gradient Descent')
ax2.set_xlabel('x1 (size)')
ax2.set_ylabel('x2 (bedrooms)')
ax2.set_zlabel('Price')

plt.show()

# --- Step 8: Print final m and b ---
print("Final slopes (m1, m2):", m)
print("Final intercept (b):", b)
print("Predictions after GD:", y_pred_final)
```
