```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


df  = pd.read_csv(r'C:/Users/mohdd/OneDrive/Documents/Desktop/GenAI/how_train_test_data_split/house_price_data.csv')
df.head()


x = df.iloc[:,:-1]
y = df["Price"]

print(f"X: \n {x}")
print(f"y: \n {y}")
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)

# This splits the dataset into two parts: training and testing
# X_train and y_train contain 80% of the data (used to train the model)
# X_test and y_test contain 20% of the data (used to test the model)
# The model is trained using X_train and y_train
# The model is evaluated using X_test and y_test

print(f"X_train: \n {x_train}")
print(f"X_test: \n {x_test}")
print(f"y_train: \n {y_train}")
print(f"y_test: \n {y_test}")
```
