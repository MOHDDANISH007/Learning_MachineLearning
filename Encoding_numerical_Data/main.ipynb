```python
import numpy as np

age = np.array([5, 12, 18, 25, 37, 45, 63]).reshape(-1, 1)
```

```python
from sklearn.preprocessing import KBinsDiscretizer



kb = KBinsDiscretizer(
    n_bins=4,
    encode='ordinal',
    strategy='uniform'
)

age_binned = kb.fit_transform(age)
print(kb.bin_edges_)
```
